## Converting a tiff stack to multiscale zarr

```python
from insegt3d.volume.io import create_multiscale_zarr_from_tiff_stack

# Data paths
path_to_tiff_stack = '/path/to/tiff_stack/'
path_to_output_zarr = 'path/to/output.zarr'

# Clipping paramaters for conversion to uint8
v_min = -0.002
v_max = 0.007

create_multiscale_zarr_from_tiff_stack(
    path_to_tiff_stack,
    path_to_output_zarr,
    dtype='uint8',
    v_min=v_min,
    v_max=v_max,
    chunks=(64,64,64),
    shards=(256,256,256)
)
```

## Saving an already processed numpy volume to a multiscale zarr

```python
import numpy as np
from insegt3d.volume.io import create_multiscale_zarr_from_numpy

path_to_output_zarr = 'path/to/output.zarr'

volume = np.random.randint(0, 256, size=(1024, 1024, 1024), dtype=np.uint8)

create_multiscale_zarr_from_numpy(
    volume,
    path_to_output_zarr,
    chunks=(64,64,64),
    shards=(256,256,256)
)
```
